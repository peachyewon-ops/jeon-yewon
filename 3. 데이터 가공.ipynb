#실행3_3_3. 변수에 조건 지정하여 새로운 그룹변수 파생

```python
#assign/pd.cut 사용
df8 = df8.assign(age_g1 = pd.cut(df8['age'],
 bins=[0, 20, 30, 40, 50], # 구간 경계 (]가 디폴트
 labels=['20대미만', '20대', '30대', '40대'], # 라벨
 right=False)) # 구간을 [) 로 바꿈
```

```python
#assign/np.select 사용
conditions = [df8['age'] >= 40,
 df8['age'] >= 30,
 df8['age'] >= 20]
choices = ['40대', '30대', '20대']
df8 = df8.assign(age_g2 = np.select(conditions, choices, default='기타')) #np.select(조건,출력, 조건이 없는 나머지)
print(df8)
```

#실행 3_4_3. 데이터 세로 결합, 가로 결합

```python
#데이터 생성하기
df_c1 = pd.DataFrame({'name': ['Kim', 'Lee', 'Park'], 'english': [90, 80, 70], 'math': [82, 75, 95]})
df_c2 = pd.DataFrame({'name': [ 'Choi', 'Song', 'Moon'], 'english': [ 40, 50, 60], 'math': [44, 68, 90]})
df_c3 = pd.DataFrame({'name': [ 'Jeong', 'Min', 'Oh'], 'english': [ 80, 50, 30], 'math': [54, 64, 88]})
df_teacher = pd.DataFrame({'class': ['c1', 'c2', 'c4'], 'teacher': ['Prof. Park', 'Prof. Lee', 'Prof. Choi']})

# 각각에 class 변수 부여
df_c1['class'] = 'c1'
df_c2['class'] = 'c2'
df_c3['class'] = 'c3'

print(df_c1)
print(df_c2)
print(df_c3)
print(df_teacher)
```

```python
# 세로로 합치기
df_all = pd.concat([df_c1, df_c2, df_c3], ignore_index=True)
#ignore_index=True: 인덱스를 무시하고 0,1,2,… 새로 생성
#ignore_index=False (기본값): 원래 DataFrame 인덱스를 유지하므로 중복
print(df_all)
```

```python
df_all = pd.concat([df_c1, df_c2, df_c3])
#ignore_index=True: 인덱스를 무시하고 0,1,2,… 새로 생성
#ignore_index=False (기본값): 원래 DataFrame 인덱스를 유지하므로 중복
print(df_all)
```

```python
#병합하기 (class 기준)
df_merge_i = pd.merge(df_all, df_teacher, on='class', how='inner')
print(df_merge_i)
df_merge_l = pd.merge(df_all, df_teacher, on='class', how='left')
print(df_merge_l)
df_merge_r = pd.merge(df_all, df_teacher, on='class', how='right')
print(df_merge_r)
df_merge_o = pd.merge(df_all, df_teacher, on='class', how='outer')
print(df_merge_o)
```
