#클래스 만들기


```python
class Rectangle: #class 이름을 지정함
  def __init__(self,width,height): #self는 Rectangle에서 만드는 자기 자신(객체) #괄호 안은 인수
     self.width=width #속성(변수)
     self.height=height
  def area(self): #메서드(함수)
    return (self.width*self.height)
```

```python
# Rectangle은 변수도 지정되고, 함수도 지정됨
r = Rectangle(3, 4) #width=3, height=4

print(r.width)
```

```python
print(r.height)
```

```python
print(r.area()) #실제 실행 Rectangle.area(r), Ractangle 아래에 area 함수가 있음
```

```python
def g_func(a,b):
  c=a+b
  d=a-b
  e=a*b
  return(c,d,e)
g_func(3,4)
```

```python
class Rectangle2:
  def __init__(self,width,height): #생성자
    self.width=width #속성(변수)
    self.height=height
  def area(self):  #매서드(함수)
    return self.width*self.height
  def perimeter(self):  #매서드(함수)
    return self.width*2+self.height*2

q=Rectangle2(5,6)
print(q.width)
print(q.height)
print(q.area())
print(q.perimeter())
```

```python
#이전에는 scipy의 sum, mean, var함수를 사용하기도 했으나 최신 버젼에서는 np만 쓰도록 정리됨
print(np.sum(score_A))
print(len(score_A)) #len(length) 기본이니까 np.를 붙이지 않음
print(np.mean(score_A))
```

##연습２_３

```python
for i in range(1, 21):
    # i를 2로 나눈 나머지가 0, 즉 i가 짝수이면
    if i % 2 == 0:
        # 현재 반복을 건너뛰고 다음 반복으로 넘어갑니다.
        continue
    # 짝수가 아니면 (홀수이면) 값을 출력합니다.
    print(i)
```

```python
total_sum = 0
for i in range(1, 21):
    if i % 2 != 0:
        total_sum = total_sum + i
print(total_sum)
```

```python
total = 0 # 합계를 저장할 변수를 초기화합니다.

for i in range(1, 21):
    # i가 짝수(i를 2로 나눈 나머지가 0)이면 건너뜁니다.
    if i % 2 == 0:
        continue

    # i가 홀수이면 total에 더합니다.
    total += i

print(f"1부터 20까지의 홀수의 합: {total}")
```
